```python
import numpy as np
import matplotlib.pyplot as plt

# --------------------------------------------------
# PHSX 216 / Acceleration Due to Gravity Lab
# Linear Least Squares Fit
# --------------------------------------------------

# DATA FROM EXCEL
# x = 2Δh [m]

x = np.array([
    0.780,
    0.740,
    0.680,
    0.566,
    0.472
])

# y = vf² [m²/s²]

y = np.array([
    7.716049,
    6.174745,
    6.046762,
    4.865917,
    4.108839
])

# uncertainty in vf² [m²/s²]

dy = np.array([
    1.067607,
    0.990352,
    0.830253,
    0.664386,
    0.558987
])


# --------------------------------------------------
# LINEAR LEAST SQUARES FIT
# Gives slope, intercept, and their uncertainties
# --------------------------------------------------

fit, covariance = np.polyfit(
    x,
    y,
    1,
    cov=True
)

slope = fit[0]
intercept = fit[1]

# Uncertainties

slope_uncertainty = np.sqrt(covariance[0,0])
intercept_uncertainty = np.sqrt(covariance[1,1])


# --------------------------------------------------
# CREATE BEST FIT LINE
# --------------------------------------------------

xnew = np.linspace(
    x.min(),
    x.max(),
    300
)

ynew = slope * xnew + intercept


# --------------------------------------------------
# CREATE GRAPH
# --------------------------------------------------

plt.figure(figsize=(10,7))

# Experimental data with error bars

plt.errorbar(
    x,
    y,
    yerr=dy,
    fmt='o',
    capsize=5,
    label='Experimental Data'
)

# Best fit line

plt.plot(
    xnew,
    ynew,
    label='Linear Best Fit'
)


# --------------------------------------------------
# LABELS
# --------------------------------------------------

plt.title('Acceleration Due to Gravity')

plt.xlabel('2Δh [m]')

plt.ylabel('$v_f^2$ [m²/s²]')

plt.grid()

plt.legend()


# --------------------------------------------------
# ANNOTATE RESULTS
# --------------------------------------------------

plt.annotate(
    'Best fit: y = mx + b',
    (0.05, 0.95),
    xycoords='axes fraction'
)

plt.annotate(
    'm = {:.4f} ± {:.4f} m/s²'.format(
        slope,
        slope_uncertainty
    ),
    (0.05, 0.90),
    xycoords='axes fraction'
)

plt.annotate(
    'b = {:.4f} ± {:.4f} m²/s²'.format(
        intercept,
        intercept_uncertainty
    ),
    (0.05, 0.85),
    xycoords='axes fraction'
)


plt.show()


# --------------------------------------------------
# PRINT FINAL RESULTS
# --------------------------------------------------

print('LINEAR FIT RESULTS')
print('----------------------------')

print(
    'Slope (g) = {:.4f} ± {:.4f} m/s²'.format(
        slope,
        slope_uncertainty
    )
)

print(
    'Intercept = {:.4f} ± {:.4f} m²/s²'.format(
        intercept,
        intercept_uncertainty
    )
)
```
